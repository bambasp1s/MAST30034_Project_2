```python
from pathlib import Path

TABLES_DIR = Path("tables")

for part in ["part_2", "part_3", "part_4"]:
    folder = TABLES_DIR / part

    print(f"\n===== {part.upper()} =====")

    for file in sorted(folder.iterdir()):
        if not file.name.startswith("."):
            print(file.name)
```

```python
import pandas as pd
from pathlib import Path

PART2_DIR = Path("tables/part_2")

part2 = pd.read_parquet(PART2_DIR)

print("Shape:", part2.shape)

print("COLUMNS")
print(part2.columns.tolist())

print("\nDATA TYPES")
print(part2.dtypes)

print("\nFIRST 10 ROWS")
display(part2.head(10))
```

```python
print("Possible ID/key columns:")

for col in part2.columns:
    if "id" in col.lower() or "abn" in col.lower():
        print(
            col,
            "| rows:", len(part2),
            "| unique:", part2[col].nunique(),
            "| missing:", part2[col].isna().sum()
        )
```
